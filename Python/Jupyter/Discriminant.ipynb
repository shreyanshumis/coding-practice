```python
import math

def solve_quadratic(a, b, c):
    if a == 0:
        return "Not a quadratic equation"
    discriminant = b**2 - 4*a*c

    if discriminant > 0:
        root1 = (-b + math.sqrt(discriminant)) / (2 * a)
        root2 = (-b - math.sqrt(discriminant)) / (2 * a)
        return f"Two distinct roots: {root1}, {root2}"
    elif discriminant == 0:
        root = -b / (2 * a)
        return f"One real root: {root}"
    else:
        return "No real roots (complex roots)"
```

```python
import networkx as nx
import matplotlib.pyplot as plt

nodes = [
    "Start",
    "Check if a=0 (Def: a)",
    "Calculate Discriminant (Def: discriminant)",
    "Discriminant > 0 (Use: discriminant)",
    "Two distinct roots (Def: root1, root2)",
    "Discriminant == 0 (Use: discriminant)",
    "One real root (Def: root)",
    "Discriminant < 0 (Use: discriminant)",
    "No real roots",
    "End"
]

edges = [
    ("Start", "Check if a=0 (Def: a)"),
    ("Check if a=0 (Def: a)", "Calculate Discriminant (Def: discriminant)"),
    ("Calculate Discriminant (Def: discriminant)", "Discriminant > 0 (Use: discriminant)"),
    ("Discriminant > 0 (Use: discriminant)", "Two distinct roots (Def: root1, root2)"),
    ("Discriminant > 0 (Use: discriminant)", "End"),
    ("Calculate Discriminant (Def: discriminant)", "Discriminant == 0 (Use: discriminant)"),
    ("Discriminant == 0 (Use: discriminant)", "One real root (Def: root)"),
    ("One real root (Def: root)", "End"),
    ("Calculate Discriminant (Def: discriminant)", "Discriminant < 0 (Use: discriminant)"),
    ("Discriminant < 0 (Use: discriminant)", "No real roots"),
    ("No real roots", "End"),
]

G = nx.DiGraph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)

plt.figure(figsize=(12, 8))
pos = nx.spring_layout(G, seed=42)
nx.draw(
    G, pos, with_labels=True, node_color="lightblue", node_size=3000, font_size=10,
    font_weight="bold", arrowsize=15
)
plt.title("Data Flow Graph for Quadratic Equation Program", fontsize=14)
plt.show()
```
