Sentence length prediction

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Generate random data for sentences and their lengths
np.random.seed(0)
num_sentences = 50
num_words_per_sentence = np.random.randint(5, 20, num_sentences)
sentence_lengths = num_words_per_sentence * 2 + np.random.normal(0, 3, num_sentences)
```

```python
# Reshape the data for sklearn compatibility
X = num_words_per_sentence.reshape(-1, 1)
y = sentence_lengths.reshape(-1, 1)
```

```python
# Train a linear regression model
model = LinearRegression()
model.fit(X, y)
```

```python
# Generate predictions for plotting
x_values = np.linspace(min(num_words_per_sentence), max(num_words_per_sentence), 100).reshape(-1, 1)
y_pred = model.predict(x_values)
```

```python
# Plot the data points and the linear regression line
plt.scatter(num_words_per_sentence, sentence_lengths, label='Data Points')
plt.plot(x_values, y_pred, color='red', label='Linear Regression')
plt.xlabel('Number of Words')
plt.ylabel('Sentence Length')
plt.title('Linear Regression on Sentence Length Prediction')
plt.legend()
plt.show()
```

Sentence splitting time

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Function to generate random sentences based on word counts
def generate_sentences(num_sentences, num_words_per_sentence_list):
    sentences = []
    for i in range(num_sentences):
        num_words_per_sentence = num_words_per_sentence_list[i]
        sentence = " ".join(np.random.choice(words, size=num_words_per_sentence))
        sentences.append(sentence)
    return sentences
```

```python
# Generate predictions for plotting
x_values = np.linspace(min(num_words_per_sentence), max(num_words_per_sentence), 100).reshape(-1, 1)
y_pred = model.predict(x_values)
```
