```python
def calculate_salary(basic_salary):
    HRA_PERCENTAGE = 0.30
    DA_PERCENTAGE = 0.80
    MA = 100
    TA = 800
    INCOME_TAX = 700
    PF = 780

    hra = basic_salary * HRA_PERCENTAGE
    da = basic_salary * DA_PERCENTAGE
    gross_salary = basic_salary + hra + da + MA + TA
    deductions = INCOME_TAX + PF
    net_salary = gross_salary - deductions

    return {
        "Basic Salary": basic_salary,
        "HRA": hra,
        "DA": da,
        "MA": MA,
        "TA": TA,
        "Gross Salary": gross_salary,
        "Deductions": deductions,
        "Net Salary": net_salary,
    }

basic_salary = float(input("Enter the basic salary of the employee: "))
salary_details = calculate_salary(basic_salary)

print("\nSalary Details:")
for key, value in salary_details.items():
    print(f"{key}: {value:.2f}")
```
