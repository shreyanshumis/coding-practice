⚡1.1 Practical - Implementation of a stochastic local search hill climbing algorithm for function optimization

```python
class NQueensCSP:
  def __init__(self, n):
    self.n = n
    self.queens = []

  def is_safe(self, row, col):
    for r,c in self.queens:
      if c == col or abs(r - row) == abs(c - col):
        return False
    return True
  def backtrack(self, row=0):
    if row == self.n:
      return True
    for col in range(self.n):
      if self.is_safe(row, col):
        self.queens.append((row, col))
        if self.backtrack(row + 1):
          return True
        self.queens.pop()
    return False

  def print_solution(self):
    board = [['.'] * self.n for _ in range(self.n)]
    for r, c in self.queens:
      board[r][c] = 'Q'
    for row in board:
      print(" ".join(row))

n = 8
n_queens = NQueensCSP(n)
if n_queens.backtrack():
  n_queens.print_solution()
else:
  print("No solution found.")
```

⚡1.3 Practical - Creation of a single layer neural network consisting of independent neurons acting on input data to produce the output

```python
import numpy as np
class SingleLayerNeuralNetwork:
  def __init__(self, input_size, output_size):
    self.weights = np.random.rand(input_size, output_size)
    self.biases = np.random.rand(output_size)
  def activation_function(self, x):
    return 1 / (1 + np.exp(-x))
  def forward(self, X):
    z = np.dot(X, self.weights) + self.biases
    output = self.activation_function(z)
    return output

if __name__ == "__main__":
  X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
  nn = SingleLayerNeuralNetwork(input_size=2, output_size=1)
  output = nn.forward(X)
  print("Output:\n", output)
```
