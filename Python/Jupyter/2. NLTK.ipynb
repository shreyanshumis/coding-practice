Normalization

```python
word2 = "Play Playing Played Apply Applied"
print(word2)
```

```python
word2.lower()
```

```python
#Tokenization
```

```python
word3 = nltk.word_tokenize(word2)
print(word3)
```

```python
#import PorterStemmer() to do Stemming
from nltk.stem import PorterStemmer
```

```python
porter = nltk.PorterStemmer()
```

```python
print(porter.stem("Play"))
print(porter.stem("Playing"))
print(porter.stem("Played"))
print(porter.stem("Apply"))
print(porter.stem("Applied"))
```

```python
[porter.stem(u) for u in word3]
```

```python
#list of comprehension
result = []
for u in word3:
  result.append(porter.stem(u))
print(result)
```

Lemmatization

```python
word2= "play playing played apply applied"
print(word2)
```

```python
#Tokenization
word3=nltk.word_tokenize(word2)
print(word3)
```

```python
#make one variabke named wln and save WordNetLemmatizer() class in it
wln = nltk.WordNetLemmatizer()
```

```python
#Stemming on list element using for loop
result = []
for w in word3:
  result.append(wln.lemmatize(w))
print(result)
```

N-Grams in NLTK

```python
sent= """my name is Amol, his name is Sachin"""
```

```python
word1=nltk.word_tokenize(sent)
print(word1)
```

```python
from nltk.collocations import BigramCollocationFinder
```

```python
find = BigramCollocationFinder.from_words(word1)
```

```python
print(find.ngram_fd.items())
```
