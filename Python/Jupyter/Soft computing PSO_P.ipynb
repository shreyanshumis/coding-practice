```python
#1
import numpy as np

# Objective function to minimize
def objective_function(x):
    return np.sum(x**2)

# PSO parameters
swarm_size = 30
dimensions = 2
iterations = 20

w = 0.5  # inertia weight
c1 = 1.5 # cognitive coefficient
c2 = 1.5 # social coefficient

x_min = -10
x_max = 10

# Initialize particle positions and velocities
positions = np.random.uniform(x_min, x_max, (swarm_size, dimensions))
velocities = np.random.uniform(-1, 1, (swarm_size, dimensions))

pBest = np.copy(positions)
pBest_scores = np.array([objective_function(p) for p in pBest])

gBest = pBest[np.argmin(pBest_scores)]
gBest_score = np.min(pBest_scores)

# Main PSO loop
for iteration in range(iterations):
    for i in range(swarm_size):
        fitness = objective_function(positions[i])

        # Update personal best
        if fitness < pBest_scores[i]:
            pBest_scores[i] = fitness
            pBest[i] = positions[i]

        # Update global best
        if fitness < gBest_score:
            gBest_score = fitness
            gBest = positions[i]

    # Update velocities and positions
    for i in range(swarm_size):
        r1, r2 = np.random.rand(2)
        velocities[i] = (
            w * velocities[i]
            + c1 * r1 * (pBest[i] - positions[i])
            + c2 * r2 * (gBest - positions[i])
        )
        positions[i] += velocities[i]

    print(f"Iteration {iteration + 1}: Best score = {gBest_score}")

print(f"Optimal solution: {gBest} with score {gBest_score}")
```

```python
#2
##Second part##
import numpy as np
import matplotlib.pyplot as plt

# Objective function
def objective_function(x):
    return np.sum(x**2)

# PSO parameters
swarm_size = 30
dimensions = 2
iterations = 20

w = 0.5  # inertia weight
c1 = 1.5  # cognitive coefficient
c2 = 1.5  # social coefficient

x_min = -10
x_max = 10

# Initialize positions and velocities
positions = np.random.uniform(x_min, x_max, (swarm_size, dimensions))
velocities = np.random.uniform(-1, 1, (swarm_size, dimensions))

pBest = np.copy(positions)
pBest_scores = np.array([objective_function(p) for p in pBest])

gBest = pBest[np.argmin(pBest_scores)]
gBest_score = np.min(pBest_scores)

# Track the best score at each iteration
best_scores = []

# Main PSO loop
for iteration in range(iterations):
    for i in range(swarm_size):
        fitness = objective_function(positions[i])

        if fitness < pBest_scores[i]:
            pBest_scores[i] = fitness
            pBest[i] = positions[i]

        if fitness < gBest_score:
            gBest_score = fitness
            gBest = positions[i]

    for i in range(swarm_size):
        r1, r2 = np.random.rand(2)
        velocities[i] = (
            w * velocities[i]
            + c1 * r1 * (pBest[i] - positions[i])
            + c2 * r2 * (gBest - positions[i])
        )
        positions[i] += velocities[i]

    best_scores.append(gBest_score)

# Plot convergence graph
plt.figure(figsize=(10, 6))
plt.plot(best_scores, label="Best Score", color="blue")
plt.title("PSO Convergence")
plt.xlabel("Iteration")
plt.ylabel("Best Score")
plt.legend()
plt.grid()
plt.show()
```
