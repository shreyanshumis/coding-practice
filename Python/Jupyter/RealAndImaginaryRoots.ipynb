```python
import math
def find_roots(a,b,c):
  #discriminant
  d = (b**2) - (4*a*c)

  #check for invalid syntax
  if(a<0 or b<0 or c<0 or a>100 or b>100 or c>100):
    print("Invalid input")
  elif a==0:
    print("Not a quadratic equation")
  elif d==0:
    root = -b/(2*a)
    print(f"roots are equal: {root:.2f}")
  elif d<0:
    real_part = -b/(2*a)
    imaginary_part = math.sqrt(-d)/(2*a)
    print(f"Imaginary roots: {real_part:.2f} + {imaginary_part:.2f}i and {real_part:.2f} - {imaginary_part:.2f}i")
  else:
    root1 = (-b + math.sqrt(d)) / (2*a)
    root2 = (-b + math.sqrt(d)) / (2*a)
    print(f"Real roots: {root1:.2f} and {root2:.2f}")


#Input section
a = int(input("Enter the value of a: "))
b = int(input("Enter the value of b: "))
c = int(input("Enter the value of c: "))

find_roots(a,b,c)
```
