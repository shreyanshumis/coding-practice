```python
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import adfuller

data = yf.download('AAPL', start='2022-01-01', end='2024-01-01')
ts = data['Close']

plt.figure(figsize=(10, 4))
plt.plot(ts, label='Apple Close Price')
plt.title('Apple Stock Price Time Series')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.show()

decomp = seasonal_decompose(ts.dropna(), model='additive', period=30)
decomp.plot()
plt.tight_layout()
plt.show()

result = adfuller(ts.dropna())
print(f"ADF Statistic: {result[0]:.4f}")
print(f"p-value: {result[1]:.4f}")
print("✅ Stationary" if result[1] < 0.05 else "⚠️ Not Stationary")
```
