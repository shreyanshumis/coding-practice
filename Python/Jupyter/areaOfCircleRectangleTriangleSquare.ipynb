```python
import math

def area_of_circle(radius):
  if radius < 0:
    return "invalid input"
  return math.pi * radius * radius

def area_of_square(side):
  if side < 0:
    return "Invalid Input"
  return side * side

def area_of_triangle(base, height):
  if base < 0 or height < 0:
    return "Invalid input"
  return 0.5 * base * height

def area_of_rectangle(length, width):
  if length < 0 or width < 0:
    return "Invalid input"
  return length * width

#Menu
print("Choose shape")
print("1 - Circle")
print("2 - Square")
print("3 - Triangle")
print("4 - Rectangle")

choice = int(input("Enter ur choice"))

if choice == 1:
  radius = float(input("Enter the radius of the circle: "))
  print(f"Area of the circle: {area_of_circle(radius):.2f}")
elif choice == 2:
  side = float(input("Enter the side length of the square:"))
  print(f"Area of the square: {area_of_square(side):.2f}")
elif choice == 3:
  base = float(input("Enter the base length of the triangle:"))
  height = float(input("Enter the height of the triangle:"))
  print(f"Area of the triangle: {area_of_triangle(base, height):.2f}")
elif choice == 4:
  length = float(input("Enter the length of the rectangle:"))
  width = float(input("Enter the width of the rectangle:"))
  print(f"Area of the rectangle: {area_of_rectangle(length, width):.2f}")
else:
  print("Invalid choice")
```
