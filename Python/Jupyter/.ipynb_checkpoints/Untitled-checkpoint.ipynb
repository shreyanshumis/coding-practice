```python
import matplotlib.pyplot as plt
```

```python
#Dataframe on BRICS countries:
data = {"country": ["Brazil","Russia","India","China","South Africa"],
        "capital": ["Brasilia","Moscow","Delhi","Beijing","Pretoria"],
        "area":[8.5,17.1,3.3,9.6,1.2],
        "population": [200,143,1252,1357,52]}

df = pd.DataFrame(data)
```

```python
#Histogram
df['population'].plot(kind='hist',bins=10)
plt.show()
```

```python
#Bar graph
df.plot(x='country', y ='population', kind='bar')
plt.show()
```

```python
#pie charts
df['population'].plot(kind='pie')
plt.show()
```

```python
#box aand whisker plot
df.boxplot(column='population')
plt.show()
```

```python
#Scatter
df.plot(x='area', y='population', kind='scatter')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset('tips')

# Bar Plot
sns.barplot(x='day', y='total_bill', data=data)
plt.title('Bar Plot')
plt.show()

# Count Plot
sns.countplot(x='day', data=data)
plt.title('Count Plot')
plt.show()

# Distribution Plot is a depricated function hence does not produce the intended output as the restr of them
sns.distplot(data['total_bill'])
plt.title('Distribution Plot')
plt.show()

# Point Plot
sns.pointplot(x='day', y='total_bill', hue='time', data=data)
plt.title('Point Plot')
plt.show()

# Facet Grid
g = sns.FacetGrid(data, col='time')
g.map(sns.histplot, 'total_bill')
g.set_axis_labels('Total Bill', 'Count')
g.set_titles(col_template='{col_name} time')
plt.show()

# Pair Grid
g = sns.PairGrid(data)
g.map(plt.scatter)
plt.suptitle('Pair Grid')
plt.show()
```
