```python
#write a program to perform a raise to power b and perform decision table testing
def raise_to_power(a,b):
  if b==0:
    return 1
  elif b<0:
    return 1/(a**abs(b))
  else:
    return a**b

a = float(input ("Enter the base number(a):"))
b = int(input("Enter the exponent (b):"))

result = raise_to_power(a,b)

print(f"{a} raised to the power of {b} is: {result}")
```

```python
from tabulate import tabulate

def raise_to_power(a,b):
  if a==0 and b<0:
    return "Undefined (0 raised to a negative power)"
  return a ** b
```

```python
test_cases = [
    (2,3),
    (3,0),
    (2,-2),
    (0,3),
    (0,0),
    (0,-1),
    (-2,2),
    (-2,3),
    (-2,0),
    (-2,-2)
]

results = [(a,b,raise_to_power(a,b)) for a, b in test_cases]

headers= ["Base (a)", "Exponent(b)", "Result"]
print(tabulate(results, headers=headers, tablefmt="grid"))
```
