```python
import networkx as nx
import matplotlib.pyplot as plt

nodes = [
    "Start",
    "Input marks for 10 students (Loop)",
    "Calculate averages (List Comprehension)",
    "Assign grades (List Comprehension)",
    "Display results (Loop)",
    "End"
]

edges = [
    ("Start", "Input marks for 10 students (Loop)"),
    ("Input marks for 10 students (Loop)", "Calculate averages (List Comprehension)"),
    ("Calculate averages (List Comprehension)", "Assign grades (List Comprehension)"),
    ("Assign grades (List Comprehension)", "Display results (Loop)"),
    ("Display results (Loop)", "End"),
    ("Input marks for 10 students (Loop)", "Input marks for 10 students (Loop)"),
    ("Display results (Loop)", "Display results (Loop)")
]

#
G = nx.DiGraph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)

adj_matrix = nx.adjacency_matrix(G).todense()

E = len(edges)
N = len(nodes)
P = 1
V_G = E - N + 2 * P

plt.figure(figsize=(10, 6))
pos = nx.spring_layout(G, seed=42)
nx.draw(
    G, pos, with_labels=True, node_color='lightgreen', node_size=3000, font_size=10,
    font_weight="bold", arrowsize=15
)
plt.title(f"Control Flow Graph (Cyclomatic Complexity V(G) = {V_G})", fontsize=14)
plt.show()

print("Adjacency Matrix:")
print(adj_matrix)
print(f"\nCyclomatic Complexity V(G) = {V_G}")
```
