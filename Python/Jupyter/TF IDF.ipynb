```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
# Example documentabs
doc = ["The burger is tasty", "The food is delicious", "The food is awful"]
# Initialize TfidfVectorizer with stop_words set to 'english'
tfidf = TfidfVectorizer(stop_words='english')
```

```python
doc_vector = tfidf.fit_transform(doc)
print(tfidf.get_feature_names_out())
df = pd.DataFrame(doc_vector.toarray(), columns=tfidf.get_feature_names_out())
print(df)
```

```python
tfidf = TfidfVectorizer(stop_words=None)
doc_vector = tfidf.fit_transform(doc)
print(tfidf.get_feature_names_out())
df = pd.DataFrame(doc_vector.toarray(), columns=tfidf.get_feature_names_out())
print(df)
```
