# Pandas Theory

Pandas is an open source data analysis lib written in py
1) power and speed of numpy to make data analysis and preprocessing
2) rich and highly robust data operations

Pandas has two types of DS:-

a) series - one dim array with indexes, it stores a single column or row of data in a dataframe.<br>
b) dataframe - tabular spreadsheet like struct representing rows each of which contains one or multiple columns.

> A one-dimensional array capable of holding any type of data - Series

> A two-dimensional data structure with columns of potentially different types iof data - Dataframe
